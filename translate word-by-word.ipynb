```python
import pickle
import numpy as np
```

```python
d = pickle.load(open("./dict_ru_en.pkl", "rb"))
en_ru_dict = d["en_ru_dict"]
ru_en_dict = d["ru_en_dict"]
```

```python
input_data = open("./En-Ru/input.txt", "r").read().splitlines()
output_data = open("./En-Ru/output.txt", "r").read().splitlines()
```

```python
def translate_word(word, dictionary):
    if word == "":
        return ""
    translations = dictionary.get(word.lower(), [word])
    translation = np.random.choice(translations)
    if word[0].isupper():
        translation = translation[0].upper() + translation[1:]
    return translation
    
```

```python
def translate(sentence, dictionary):
    for c in list(".,'\"!?"):
        sentence = sentence.replace(c, " " + c)
    sentence = sentence.replace("-", " - ")
        
    translation = " ".join([translate_word(word, dictionary) for word in sentence.split(" ")])
    for c in list(".,'\"!?"):
        translation = translation.replace(" " + c, c)
    sentence = sentence.replace(" - ", "-")
    
    return translation
```

```python
output_base = []
for sentence in input_data:
    output_base.append(translate(sentence, en_ru_dict))
open("base_output.txt", "w").write('\n'.join(output_base))
```

```python
input_base = []
for sentence in output_data:
    input_base.append(translate(sentence, ru_en_dict))
open("base_input.txt", "w").write('\n'.join(input_base))
```
