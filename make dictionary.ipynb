```python
from collections import defaultdict
```

```python
f = open("./En-Ru/en-ru-full.txt", "r")

file = f.read()

en_ru_dict = defaultdict(list)

for en, ru in map(lambda x: x.split(" "), file.splitlines()):
    en_ru_dict[en].append(ru)    
```

```python
f = open("./En-Ru/ru-en-full.txt", "r")

file = f.read()

ru_en_dict = defaultdict(list)

for ru, en in map(lambda x: x.split(" "), file.splitlines()):
    ru_en_dict[ru].append(en)    
```

```python
len(en_ru_dict), len(ru_en_dict)
```

```python
import pickle
```

```python
pickle.dump({"en_ru_dict": en_ru_dict, "ru_en_dict": ru_en_dict},
          open("dict_ru_en.pkl", "wb"))
```
